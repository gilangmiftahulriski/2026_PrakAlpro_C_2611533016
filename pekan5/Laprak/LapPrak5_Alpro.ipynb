## 2.1 Dasar Teori
Pada praktikum ini mahasiswa melakukan pengenalan dan mengimplementasikan tentang perulangan for dan nested for.

1. Perulangan. Perulanngan atau loop digunakan untuk mengeksekusi blok kode berulang kali hingga suatu kondisi terpenuhi atau semua item dalam suatu urutan diproses. Jenis utamanya adalah for loop (mengulangi urutan) dan While loop (mengeksekusi kode berdasarkan suatu kondisi).
2. Perulangan for. Perulangan for digunakan untuk mengulangi suatu urutan (yaitu daftar, tupel, kamus, himpunan, atau string). Dengan perulangan for kita dapat mengeksekusi sekumpulan pernyataan, satu kali untuk setiap item dalam daftar, tupel, kumpulan, dll. Kita bisa menggunakanny dengan menggunakan pernyataan for ... in range (....)
3. Nested for loop. Bahasa pemrograman Python memungkinkan penggunaan satu lingkaran di dalam loop lain. Bagian berikut menunjukkan beberapa contoh untuk menggambarkan konsep tersebut. Kita membuat nested for dengan menggunakan perulangan for dalam perulangan for.
4. range(start, stop) stop adalah batas akhir yang tidak diikutsertakan. Jadi perulangannya sebenarnya berhenti ketika mencapai nilai stop, namun nilai terakhir yang dijalankan/dicetak adalah stop - 1.

- 1 Nilai range(stop): Jika hanya diisi 1 nilai, nilai tersebut dianggap sebagai stop. Perulangan otomatis dimulai dari 0 dan angka terakhir yang diproses adalah stop - 1

- 2 Nilai range(start, stop): Jika diisi 2 nilai, nilai pertama adalah start (titik awal) dan nilai kedua adalah stop (batas akhir). Perulangan dimulai dari start hingga angka terakhir stop - 1.

### 2.2.5  Program nested_for1_2611533016
