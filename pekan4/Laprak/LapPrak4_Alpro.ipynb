## 2.1 Dasar Teori
Conditional merupakan salah satu struktur dalam pemrograman Python yang digunakan untuk menjalankan suatu perintah berdasarkan kondisi tertentu. Dengan menggunakan conditional, program dapat menentukan apakah sebuah perintah akan dijalankan atau dilewati berdasarkan hasil evaluasi dari suatu kondisi.

1. Pernyataan if digunakan untuk menjalankan perintah apabila kondisi yang diberikan memiliki nilai benar (True). Contohnya, jika kondisi umur == 18 bernilai benar, maka perintah yang berada di dalam blok if akan dijalankan.
2. Pernyataan if-else digunakan ketika terdapat dua kemungkinan hasil dari suatu kondisi. Jika kondisi pada if bernilai benar, maka perintah pada bagian if akan dijalankan. Sebaliknya, apabila kondisi tersebut salah (False), program akan menjalankan perintah pada bagian else.
3. Pernyataan if-elif-else digunakan apabila terdapat lebih dari dua kondisi yang harus diperiksa. Python akan mengevaluasi kondisi secara berurutan mulai dari if, kemudian elif, hingga else, dan menjalankan perintah berdasarkan kondisi yang terpenuhi atau bernilai benar.

### 2.2.4  Program multi_if2_2611533016

code:

````python
# Buat file dengan nama multi_if2_NIM.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_3016
# Program ini menggunakan fungsi input()
# Program menghitung diskon belanja

# Input dari user
total_belanja_3016 = float(input("Masukkan total belanja (Rp):  "))

# Input status member (mengecek apakah user mengetik 'y' atau ''ya')
input_member_3016 = input("Apakah anda member?(y/t): ").strip().lower()
is_member_3016 = input_member_3016 in ["y" , "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_3016 = input("Apakah kode promo valid ? (y/t). ").strip().lower()
kode_promo_valid_3016 = input_promo_3016 in ["y" , "ya"]

total_diskon_persen_3016 = 0

# Multi-IF terpisah: setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memnuhi beberap syarat sekaligus

if total_belanja_3016 > 1000000:
    total_diskon_persen_3016 += 10 # Diskon belanja besar

if is_member_3016:
    total_diskon_persen_3016 += 5 # Diskon member

if kode_promo_valid_3016:
    total_diskon_persen_3016 += 15 # Diskon voucher

# menghitung nominal diskon dan total bayar 
nominal_diskon_3016 = total_belanja_3016 * (total_diskon_persen_3016 / 100)
total_bayar_3016 = total_belanja_3016 - nominal_diskon_3016
# output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : (total_diskon_persen)%(Rp {nominal_diskon_3016:,.0f})")
print(f"Total bayar : Rp {total_bayar_3016 :,.0f}")

print(f"total diskon yang anda dapatkan: {total_diskon_persen_3016}%")
# output : Total diskon yang anda dapatkan : 30% jika belanja > 1 juta, member, dan kode promo valid
````

5. Setelah itu, program meminta pengguna memasukkan kode promo dengan pilihan “y” atau “t”. Input tersebut diproses dengan cara yang sama dan hasilnya disimpan pada variabel kode_promo_valid_3016.
6. Variabel total_diskon_persen_3016 diberikan nilai awal 0. Variabel ini digunakan untuk menyimpan besarnya persentase diskon yang diperoleh.

12. Selanjutnya, program menentukan total pembayaran dengan cara mengurangi jumlah total belanja dengan nominal diskon.
13. Program menampilkan rincian pembayaran, yaitu persentase diskon, jumlah nominal diskon dalam rupiah, serta total pembayaran yang harus dibayarkan.
14. Terakhir, program kembali menampilkan jumlah diskon yang diperoleh dalam bentuk persentase sebagai hasil akhir dari proses perhitungan.

### 2.2.7  Program match_case1_2611533016

code:

````python
# Buat file fengan nama match-case1_Nim.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_3016
# Program ini menggunakan fungsi input()
# program konversi bilai angka menjadi nama bulan

bulan_3016 = int(input("Masukkan angka bulan (1-12): "))

match bulan_3016:
    case 1: 
        print("Januari")
    case 2: 
        print("Februari")
    case 3: 
        print("Maret") 
    case 4: 
        print("April")
    case 5: 
        print("Mei")
    case 6: 
        print("Juni")
    case 7: 
        print("Juli")
    case 8: 
        print("Agustus")
    case 9: 
        print("September")
    case 10: 
        print("Oktober")
    case 11: 
        print("November")
    case 12: 
        print("Desember")
    case _:
        print("Angka tidak valid")
````

hasil:

```
Masukkan angka bulan (1-12): 9
September
```

Penjelasan:

1. Program diawali dengan meminta pengguna memasukkan angka bulan dari 1 sampai 12 menggunakan fungsi input().
2. Angka yang dimasukkan kemudian dikonversi menjadi tipe data integer dan disimpan dalam variabel bulan_3016.
3. Program menggunakan struktur match-case untuk mencocokkan nilai yang dimasukkan dengan beberapa pilihan bulan yang tersedia.
4. Jika angka yang dimasukkan adalah 1, program mencocokkannya dengan case yang sesuai dan menampilkan nama bulan Januari.
5. Untuk nilai 2 sampai 12, program akan menyesuaikan angka tersebut dengan case masing-masing dan menampilkan nama bulan yang sesuai, mulai dari Februari hingga Desember.
6. Apabila angka yang dimasukkan berada di luar rentang 1 sampai 12, tidak ada case yang sesuai sehingga program menjalankan bagian default dan menampilkan pesan “Angka tidak valid”.
7. Setelah nama bulan atau pesan kesalahan ditampilkan, program berakhir dan menyelesaikan prosesnya.
