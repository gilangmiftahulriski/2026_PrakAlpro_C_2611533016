## 2.1 Dasar Teori
Pada praktikum ini mahasiswa melakukan pengenalan dan mengimplementasikan tentang variabel, tipe data (seperti string, boolean, integer, float, dan complex, ), konversi tipe data, konstanta, hingga sistem input dari pengguna

1. Tipe Data Numerik

ada python terdapat beberapa tipe data numerik yaitu integer untuk bilangan bulat (100), tipe data float untuk bilangan pecahan atau desimal (10.2345), dan ada juga tipe data complex untuk bilangan imajiner atau kompex (100+3j).

2. Tipe data string.

Tipe data string (str) adalah tipe data untuk menyimpan teks atau karakter. Pada python dapat digunakan dengan menggunakan tanda peting tunggal ('), petik ganda ("), petik tiga (""") untuk teks baris lebih dari 1.

3. Tipe Data Boolean.

Tipe data boolean adalah tipe data yang hanya mengeluarkan True dan False. Tipe ini biasanya digunakan untuk melakukan perbandingan terhadap sesuatu sehingga dibutuhkan hasil apakah dia true atau false.

4. Fungsi input dan konversi data.

a. input().

FUngsi input() digunakan untuk memungkinkan program untuk mendapatkan masukkan dari pengguna. fungsi input ini akan menghasilkan nilai yang bertipe data string (str).

b. Konversi tipe data.

Fungsi ini biasanya digunakan untuk mengubah tipe data yang diberikan oleh pengguna "input()" yang memberikan nilai bertipe string. Untuk mengubah tipe data kita dapat menggunakan " int() " untuk mengubah tipe data menjadi integer (bilangan bulat), " float() " untuk menjadikannya bilangan desimal, atau bisa juga menggunakan " str() " jika kita ingin mengubah suatu tipr data menjadi string.

5. Variabel.

Variabel adalah sebuah wadah tempat di mana kita bisa memasukkan sesuatu di dalamnya, yaitu memasukkan atau menyimpan data. Untuk menyimpan nilai ke dalam variabel kita perlu menggunakan tanda "=" . contoh " panjang_3016 = 30 ".

6. Konstanta.

Konstanta yaitu variabel yang nilainya tidak boleh berubah. contohnya pada " PI_3016: Final = 3.14 " ini menandai bahwa nilai dari PI tidak boleh berubah yg bernilai 3.14.

Pada praktikum ini, kode python dengan file bernama Numerik_2611533016 digunakan untuk mengetahui tipe data numerik dari 3 tipe data yaitu integer, float, dan complex.

````python
# Buat file dengan nama Numerik_2611533016.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_3016
#create a variabel with numeric type value.
a_3016=100
print("The type of variable having value", a_3016, "is", type (a_3016))

#create a variabel with float value.
b_3016=10.2345
print("The type of variable having value", b_3016, "is", type (b_3016))

#create a variabel with complex value.
c_3016=100+3j
print("The type of variable having value", c_3016, "is", type (c_3016))
````

````python
# Buat file dengan nama Boolean_NIM.py
# nama variabel ditambah 4 digit nim terakhir contoh: nilai_3016
# Deklarasi variabel dengan tipe data Boolean
is_lulus_3016 = True
is_cumlaude_3016 = True

# Menggunakan Boolean
nilai_3016 = 85
batas_lulus_3016 = 75

# Menenentukan nilai Boolean dari kondisi
status_kelulusan_3016 = nilai_3016 >= batas_lulus_3016 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai : ", nilai_3016)
print("Apakah Lulus? : ", status_kelulusan_3016)
if is_lulus_3016 and is_cumlaude_3016:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
````

### 2.2.5  Program Konstanta_2611533016

Program menghitung luas lingkaran dengan menetapkan PI sebagai konstanta yang tidak bisa diubah.

code:

````python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh:  jari_3016

from typing import Final
PI_3016: Final = 3.14
print("pi: %f" % (PI_3016))
jari_3016 = float(input("Masukkan nilai jari-jari: "))
luas_3016 = PI_3016 * jari_3016 * jari_3016
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3016, luas_3016))
````

hasil:

```

```

Penjelasan:

1. Mengimpor Final dari modul typing untuk menetapkan variabel PI_1016 bernilai 3.14 sebagai konstanta.
2. Tanda %f berarti angka yang ditampilkan berubah menjadi desimal dengan 6 angka dibelakang koma.
3. Variabel jari_3016 menerima input jari jari dari pengguna dan dikonversi menjadi tipe data float.
