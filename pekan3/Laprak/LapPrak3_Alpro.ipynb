## 2.1 Dasar Teori
Python memiliki berbagai jenis operator yang digunakan untuk melakukan proses tertentu terhadap data dan variabel. Operator merupakan simbol atau kata khusus yang digunakan untuk melakukan operasi, seperti perhitungan, perbandingan, pemberian nilai, maupun pengolahan kondisi. Penggunaan operator menjadi salah satu dasar penting dalam pemrograman karena hampir setiap program membutuhkan operasi terhadap data.

1. Operator Aritmatika Operator aritmatika digunakan untuk melakukan operasi perhitungan matematika pada Python. Operator yang termasuk di dalamnya adalah penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat (//), modulus atau sisa pembagian (%), dan perpangkatan (**). Operator aritmatika dapat digunakan pada berbagai tipe data numerik, seperti integer dan float. Penggunaannya banyak ditemukan dalam perhitungan nilai, transaksi, luas bangun, dan berbagai proses perhitungan lainnya.
2. Operator Perbandingan Operator perbandingan digunakan untuk membandingkan dua nilai atau lebih. Hasil dari operasi perbandingan berupa nilai Boolean, yaitu True atau False. Operator yang digunakan meliputi sama dengan (==), tidak sama dengan (!=), lebih besar (>), lebih kecil (<), lebih besar atau sama dengan (>=), serta lebih kecil atau sama dengan (<=). Operator ini biasanya digunakan dalam percabangan dan pemeriksaan kondisi pada suatu program.
3. Operator Logika Operator logika digunakan untuk menggabungkan atau memanipulasi beberapa kondisi. Python memiliki tiga operator logika utama, yaitu and, or, dan not. Operator and menghasilkan nilai benar apabila semua kondisi yang diperiksa bernilai benar, sedangkan or menghasilkan nilai benar apabila terdapat setidaknya satu kondisi yang benar. Sementara itu, not digunakan untuk membalik nilai Boolean dari suatu kondisi. Operator logika banyak digunakan dalam proses pengambilan keputusan pada program.
4. Operator Penugasan Operator penugasan digunakan untuk memberikan nilai kepada variabel atau mengubah nilai yang telah tersimpan di dalam variabel. Operator yang paling dasar adalah =, sedangkan operator gabungan lainnya antara lain +=, -=, *=, /=, //=, %= dan **=. Operator tersebut dapat membuat penulisan program menjadi lebih sederhana, terutama ketika suatu variabel perlu diperbarui berdasarkan nilai sebelumnya.
5. Operator Bitwise Operator bitwise digunakan untuk melakukan operasi pada tingkat bit dari suatu bilangan integer. Python menyediakan beberapa operator bitwise, yaitu AND (&), OR (|), XOR (^), NOT (~), left shift (<<), dan right shift (>>). Operasi ini bekerja dengan memproses representasi biner dari suatu bilangan. Operator bitwise biasanya digunakan dalam pengolahan data tingkat rendah, manipulasi bit, sistem komputer, dan beberapa kebutuhan pemrograman tertentu.
6. perator Keanggotaan Operator keanggotaan digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah objek seperti string, list, tuple, atau set. Python memiliki operator in dan not in. Operator in menghasilkan True apabila suatu nilai ditemukan di dalam objek, sedangkan not in menghasilkan True apabila nilai tersebut tidak ditemukan. Operator ini dapat digunakan untuk melakukan pemeriksaan data dengan cara yang sederhana.
7. Operator Identitas Operator identitas digunakan untuk mengetahui apakah dua variabel mengacu pada objek yang sama di dalam memori. Python menyediakan operator is dan is not. Operator is menghasilkan True apabila kedua variabel mengacu pada objek yang sama, sedangkan is not menghasilkan True apabila keduanya tidak mengacu pada objek yang sama. Operator identitas berbeda dengan operator == yang digunakan untuk membandingkan nilai.

### 2.2.2  Program assignment_2611533016

````python
# Buat file dengan nama assignment_2611533016.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_3016
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Progarm operator assignment dalam python

angka1_3016 = int(input("Input angka-1: "))
angka2_3016 = int(input("Input angka-2: "))

print("nilai awal angka1 =", angka1_3016)
print("nilai angka2 =", angka2_3016)

# Assignment biasa
hasil_3016 = angka1_3016
print("\nAssignment Biasa (=)")
print("Hasil =", hasil_3016)

# Assignment penambahan
hasil_3016 = angka1_3016
hasil_3016 += angka2_3016
print("\nAssignment Penambahan (+=)")
print("Hasil =", hasil_3016)

# Assignment pengurangan
hasil_3016 = angka1_3016
hasil_3016 -= angka2_3016
print("\nAssignment Pengurangan (-=)")
print("Hasil =", hasil_3016)

# Assignmet perkalian
hasil_3016 = angka1_3016
hasil_3016 *= angka2_3016
print("\nAssignment Perkalian (*=)")
print("Hasil =", hasil_3016)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_3016 != 0:
    hasil_3016 = angka1_3016
    hasil_3016 /= angka2_3016
    print("\nAssignment Pembagian (/=)")
    print("Hasil =", hasil_3016)
    #Operator tambahan
    hasil_3016 = angka1_3016
    hasil_3016 //= angka2_3016
    print("\nAssignment Pembagian Bulat (//=)")
    print("Hasil =", hasil_3016)
    hasil_3016 = angka1_3016
    hasil_3016 %= angka2_3016
    print("\nAssignment Sisa Bagi (%=)")
    print("Hasil =", hasil_3016)
else:
    print("Pembagian tidak dapat dilakukan")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_3016 = angka1_3016
hasil_3016 **= angka2_3016
print("\nAssignment Perpangkatan (**=)")
print("Hasil =", hasil_3016)
````

### 2.2.3  Program perbandingan_2611533016

````python
# Buat file dengan nama Perbandingan_2611533016.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka_3016
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Progarm operator perbandingan dalam python

angka1_3016 = int(input("Input angka-1: "))
angka2_3016 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_3016 = angka1_3016 > angka2_3016
print("\nOperator Lebih Besar Dari")
print("angka1 > angka2 =", hasil_3016)

# Lebih kecil dari
hasil_3016 = angka1_3016 < angka2_3016  
print("\nOperator Lebih Kecil Dari")
print("angka1 < angka2 =", hasil_3016)

# Lebih besar dari sama dengan
hasil_3016 = angka1_3016 >= angka2_3016
print("\nOperator Lebih Besar Dari atau Sama Dengan")
print("angka1 >= angka2 =", hasil_3016)

# Lebih kecil dari sama dengan  
hasil_3016 = angka1_3016 <= angka2_3016
print("\nOperator Lebih Kecil Dari atau Sama Dengan")
print("angka1 <= angka2 =", hasil_3016)

# Sama dengan
hasil_3016 = angka1_3016 == angka2_3016
print("\nOperator Sama Dengan")
print("angka1 == angka2 =", hasil_3016)

# Tidak sama dengan
hasil_3016 = angka1_3016 != angka2_3016
print("\nOperator Tidak Sama Dengan")
print("angka1 != angka2 =", hasil_3016)

# Tambahan perbandingan berantai dalam Python
hasil_3016 = 0 < angka1_3016 < 100
print("\nOperator Perbandingan Berantai")
print("0 < angka1 < 100 =", hasil_3016)

hasil = 0 < angka2_3016 < 100
print("0 < angka2 < 100 =", hasil)
````

### 2.2.5  Program bitwise_2611533016

code:

````python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n========================================")
print("3. OPERATOR BITWISE")
print("========================================")

angka1_3016 = int(input("Masukkan angka bitwise-1: "))
angka2_3016 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_3016, "| biner =", bin(angka1_3016))
print("angka2 =", angka2_3016, "| biner =", bin(angka2_3016))

# Bitwise AND
hasil_3016 = angka1_3016 & angka2_3016
print("\nBitwise AND (&)")
print(angka1_3016, "&", angka2_3016, "=", hasil_3016)
print("Biner hasil =", bin(hasil_3016))
print("Biner hasil (8 bit) =", format(hasil_3016    , "08b"))

# Bitwise OR
hasil_3016 = angka1_3016 | angka2_3016
print("\nBitwise OR (|)")
print(angka1_3016, "|", angka2_3016, "=", hasil_3016)
print("Biner hasil =", bin(hasil_3016))
print("Biner hasil (8 bit) =", format(hasil_3016, "08b"))

# Bitwise XOR
hasil_3016 = angka1_3016 ^ angka2_3016
print("\nBitwise XOR (^)")
print(angka1_3016, "^", angka2_3016, "=", hasil_3016)
print("Biner hasil =", bin(hasil_3016))
print("Biner hasil (8 bit) =", format(hasil_3016, "08b"))

# Bitwise NOT
hasil_3016 = ~angka1_3016
print("\nBitwise NOT (~)")
print("~", angka1_3016, "=", hasil_3016)
print("Biner hasil =", bin(hasil_3016))
print("Biner hasil (8 bit) =", format(hasil_3016, "08b"))

# Bitwise geser kiri
jumlah_geser_3016 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_3016 = angka1_3016 << jumlah_geser_3016
print("\nBitwise geser kiri (<<)")
print(angka1_3016, "<<", jumlah_geser_3016, "=", hasil_3016)
print("Biner hasil =", bin(hasil_3016))
print("Biner hasil (8 bit) =", format(hasil_3016, "08b"))

# Bitwise geser kanan
hasil_3016 = angka1_3016 >> jumlah_geser_3016
print("\nBitwise geser kanan (>>)")
print(angka1_3016, ">>", jumlah_geser_3016, "=", hasil_3016)
print("Biner hasil =", bin(hasil_3016))
print("Biner hasil (8 bit) =", format(hasil_3016, "08b"))
````

````python
print("=== SISTEM TRANSAKSI TOKO ===")
nama_3016 = input("Masukkan Nama Pelanggan : ")
status_3016 = input("Masukkan Status Pelanggan (member/nonmember) : ")
total_3016 = int(input("Masukkan Total Belanja : "))
jumlah_3016 = int(input("Masukkan Jumlah Barang : "))
promo_3016 = input("Masukkan Kode Promo : ")

print("\n=== DATA TRANSAKSI ===")
print(f"Nama Pelanggan       : {nama_3016}")
print(f"Status Pelanggan     : {status_3016}")
print(f"Total Belanja        : Rp{total_3016}")
print(f"Jumlah Barang        : {jumlah_3016}")
print(f"Kode Promo           : {promo_3016}")

kode_promo_3016 = ["HEMAT10", "HEMAT20", "HIDUPMAHASISWA", "HIDUPFTI"] #list kode promo

syarat_total_3016 = total_3016 >= 200000 # apakah memenuhi syarat belanja
syarat_jumlah_3016 = jumlah_3016 >= 3 # apakah memenuhi syarat barang
status_valid_3016 = status_3016 == "member" # apakah user member
promo_valid_3016 = promo_3016 in kode_promo_3016 # apakah kode promo ada dalam list

print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000        : {syarat_total_3016}")
print(f"Jumlah Barang >= 3         : {syarat_jumlah_3016}")
print(f"Status Member              : {status_valid_3016}")
print(f"Kode Promo Tersedia        : {promo_valid_3016}")
print(f"Mendapatkan Diskon         : {syarat_jumlah_3016 or syarat_total_3016}")
print(f"Mendapatkan Promo          : {promo_valid_3016}")

diskon_3016 = 0.05

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                      : Rp{diskon_3016 * total_3016}")
print(f"Total Pembayaran            : Rp{total_3016 - total_3016 * diskon_3016}")
print(f"Rata-rata Harga Barang      : Rp{(total_3016 - total_3016 * diskon_3016)/jumlah_3016}")

print("\n=== HAK AKSES PELANGGAN ===")
print(f"Kode Hak Akses              : ...")
print(f"Member Access               : {status_valid_3016}")
print(f"Promo Access                : {promo_valid_3016}")
print(f"Free Shipping Access        : ...")

kode_transaksi_3016 = int(status_valid_3016) << 0 | int(syarat_total_3016) << 1 | int(syarat_jumlah_3016) << 2 | int(promo_valid_3016) << 3
kode_referensi_3016 = int(status_valid_3016) << 0 | int(syarat_total_3016) << 1 | int(promo_valid_3016) << 3

print("\n=== OPERASI BITWISE ===")
print("=== Kode Status Transaksi ===")
print(f"{format(int(status_valid_3016) << 0,'04b')} | {format(int(syarat_total_3016) << 1,'04b')} | {format(int(syarat_jumlah_3016) << 2,'04b')} | {format(int(promo_valid_3016) << 3,'04b')}")
print(f"Kode Biner   : {format(kode_transaksi_3016,'04b')}")
print(f"Kode Desimal : {kode_transaksi_3016}")

print("\n=== Pemeriksaan Status ===")
print("Cek Member")
print(f"{format(kode_transaksi_3016,'04b')} & {format(int(status_valid_3016) << 0,'04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_3016) & int(status_valid_3016) << 0,'04b')}")
print(f"Hasil Desimal : {(kode_transaksi_3016) & int(status_valid_3016) << 0}")

print("Cek Promo")
print(f"{format(kode_transaksi_3016,'04b')} & {format(int(promo_valid_3016) << 3,'04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_3016) & int(promo_valid_3016) << 3,'04b')}")
print(f"Hasil Desimal : {(kode_transaksi_3016) & int(promo_valid_3016) << 3}")

print("\n=== Perbandingan Status ===")
print(f"Kode Transaksi : {format(kode_transaksi_3016,'04b')}")
print(f"Kode Referensi : {format(kode_referensi_3016,'04b')}")
print(f"{format(kode_transaksi_3016,'04b')} ^ {format(kode_referensi_3016,'04b')}")
print(f"Hasil Biner   : {format((kode_transaksi_3016) ^ (kode_referensi_3016),'04b')}")
print(f"Hasil Desimal : {(kode_transaksi_3016) ^ (kode_referensi_3016)}")

print("\n=== Shift ===")
print(f"{format(kode_transaksi_3016,'04b')} << 1")
print(f"Hasil Biner   : {format((kode_transaksi_3016) << 1,'04b')}")
print(f"Hasil Desimal : {(kode_transaksi_3016) << 1}") 
print("=== SELESAI ===")
````
